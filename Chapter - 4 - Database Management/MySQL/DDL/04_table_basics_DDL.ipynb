## Table Basics
A relational database system contains one or more objects called tables. The data or information for the database are stored in these tables. Tables are uniquely identified by their names and are compromised of columns and rows.

## Create table

- create table employee(
* emp_id int not null,
* emp_name varchar(255),
* manager_name varchar(255),
* divion int,
* primary key  (emp_id)
* );
