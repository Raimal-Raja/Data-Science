## What are constraints?
When tables are created, it  is common for one or more columns to have constraints associated with them. A constraints is basically a rule associated with a column that the data entered into that column must follow.
- for example; a 'unique' constraints specifies that no two records can have the same value in a particular colunn. They must all be unique.
- The other two most popular constraints are 'not null' which specifies that a column can't be left blank, and 'primary key' a 'primary key' constraint defines a unique idenitification of  each records (or row) in a table.
