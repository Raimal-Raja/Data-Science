### Feature Binning
Feature bining converts or transform continous/numeric variable to categorical variable. It can also be used to identify missing values or outliers.

- Equal width binning
- Equal frequency binning

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
from sklearn import preprocessing
```

```python
df = pd.read_csv("D:\GitHub\Data-Science\datasets\Churn_Modelling.csv")
```

```python
df.drop(columns=['CustomerId','RowNumber','Surname'],axis=1,inplace=True)
```

```python
df.head(5)
```

```python
labels = ['0-20','21-40','41-60','Above 60']
bins = [0,21,40,61,100]

df['Age_bins'] = pd.cut(df.Age, bins, labels=labels, include_lowest=True)
```

```python
df.head(5)
```

```python
plt.bar(labels,df.Age_bins.value_counts())
plt.title('Age Count')

plt.xlabel('Age bins')
plt.ylabel('Age Count')
plt.show()
```

```python
def add_labels(x,y):
    for i in range(len(x)):
        plt.text(i,y[i],y[i])
```

```python
plt.bar(labels,df.Age_bins.value_counts())

add_labels(labels,df.Age_bins.value_counts())
plt.title('Age Count')

plt.xlabel('Age bins')
plt.ylabel('Age Count')
plt.show()
```
