```python
import numpy as np
import pandas as pd
import matplotlib.ticker as mtick
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

### Univariate Analysis

```python
df = pd.read_csv('D:\GitHub\Data-Science\datasets\Churn_Modelling.csv')
```

```python
df.head(4)
```

```python
telco = df[['Gender','Geography','Exited']]
```

```python
telco
```

```python
for  i, predictor in enumerate(telco.drop(columns=['Exited'])):
    plt.figure()
    sns.countplot(data=telco, x=predictor, hue='Exited')
```

### Bivariate Analysis

```python
sns.histplot(x='Gender',hue='Geography',data=telco,stat='count',multiple='dodge')
plt.show()
```

```python
telco_target = telco.loc[telco['Exited']==1]
```

```python
sns.histplot(x='Gender',hue='Geography',data=telco_target,stat='count',multiple='dodge')
```

### Numerical Analysis

### correlation: It range from -1 to +1
    - +1 -> Positive Correlation
    - -1 -> Negative Correlation
    - ~0 -> No Correlation

```python
Numerical_data = df[['RowNumber','Age','Tenure']]
```

```python
Numerical_data.corr()
```

```python
plt.figure(figsize=(20,8))
Numerical_data.corr()['Age'].sort_values(ascending=False).plot(kind='bar')
```
