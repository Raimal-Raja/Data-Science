Log Loss (short for Logarithmic Loss), also known as Cross-Entropy Loss, is a performance metric used to evaluate classification models—specifically those that output probabilities rather than just hard labels (like 0 or 1).

While metrics like Accuracy or Precision only care about whether the model got the answer right or wrong, Log Loss cares about how confident the model was in its prediction.

How It Works
Log Loss measures the "closeness" of a model's predicted probabilities to the actual true labels.

The Goal: You want a Log Loss score of 0, which represents perfect prediction.

The Penalty: The metric heavily penalizes models that are confident and wrong.

A Quick Example
Imagine a football match where Team A actually wins (True Label = 1).

Model 1 predicts Team A has a 90% chance of winning. (Log Loss is very low/good).

Model 2 predicts Team A has a 51% chance of winning. (Log Loss is higher, because it was unsure).

Model 3 predicts Team A has a 1% chance of winning. (Log Loss is extremely high/bad because the model was highly confident in the wrong answer).
