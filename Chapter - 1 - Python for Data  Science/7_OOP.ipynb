In Python object oriented programming is a programming paradigm that uses objects and clases in programming. it aims to implement real-world entities like inheritance, polymorphisms, encapsulation, etc. in the programming. the main concept of OOP is to bind the data and the functions that work on that together as a single unit  so taht no other part of the code can access this data

### Main concept of object oriented programming
**1.Class**
**2.Object**
**3.polymorphism**
**4.encapsulation**
**5.Inheritance**
