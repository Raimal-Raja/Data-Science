## Additional Rule of Probability

### Rule -1
when two event A and B are mutually exclusive, the probability that  A or B will occur is the sum of probability of each event

P(AuB) = P(A)+ P(B)

### Rule -2
When two event A  and B not mutually exclusive, the probability that A or B will occur is the sum of probability of each event and minus of intersection of A and B

P(AuB) = P(A)+P(B)-P(AnB)
