## Types of sampling
- Probability sampling
- Non-Probability Sampling


### - Probability sampling
- Every member of the population has an equal chance of being selected

- Examples:
    - Simple random sampling
    - Stratified random sampling
    - Cluster sampling
    - Systematic sampling


### - Non-Probability sampling
- Sampling are selected on the basis of judgement or the convenience of accessing data

- Largely depends on the researcher's sample selection skills

- Example:
    - Conveniece sampling
    - purposive sampling
    - voluntary response sampling
    - snowball sampling
