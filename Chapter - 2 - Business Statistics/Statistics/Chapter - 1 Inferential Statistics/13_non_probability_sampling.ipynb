### Convenience Sampling
- Include the respondents/member who are easy to reach for  researcher.

### Purposive Sampling
- Select a sampling based on  the purpose of  the research.
- Researcher select the sample by using their expertise and knowledge.

### Voluntary  Response Samping
- Based on the ease of access.
- Members volunteer themselves instead of research selecting the participants and directly contracting them.

## Snowball Sampling
- Recruit the participants via research participants for test or study
- used where it's hard to find find the potential population for research.
