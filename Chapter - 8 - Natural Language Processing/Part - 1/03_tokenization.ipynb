```python
corpus = """
Hello Welcome, This is Data Scientist, Raimal Raja.
Please! rate my linkedn Profile.
"""
```

```python
print(corpus)
```

```python
## Tokenization - 1
## Sentence --> paragraphs
import nltk
nltk.download('punkt', force=True)

from nltk.tokenize import sent_tokenize

```

```python
import nltk
nltk.data.find('tokenizers/punkt/english.pickle')
```

```python

document = sent_tokenize(corpus, language='english')
```

```python
document = sent_tokenize(corpus)
```

```python
type(document)
```

```python
for sentence in document:
    print(sentence)
```

```python
# Tokenization - 2
# paragraph --> words
# Sentence -- > Words

from nltk.tokenize import word_tokenize
```

```python
words = word_tokenize(corpus)
```

```python
for sentence in document:
    print(word_tokenize(sentence))
```

```python
## this treat punctuation as single word
from nltk.tokenize import wordpunct_tokenize
```

```python
word_punc = wordpunct_tokenize(corpus)
print(word_punc)
```

```python
# full stop will not be treated as a saparate word 
# but last full stop of that paragraph will be treated as single saparate word
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()

words = tokenizer.tokenize(corpus)

print(words)
```
