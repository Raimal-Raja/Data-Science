```python
import nltk
nltk.download('maxent_ne_chunker')
```

```python
sentence = 'The Eiffel Tower was built from 1887 to 1889 by a French Engineer Gautav Eiffel'
```

```python
from nltk import word_tokenize,pos_tag

word = word_tokenize(sentence)
```

```python
word
```

```python
postag = pos_tag(word)
```

```python
postag
```

```python
nltk.download('words')
```

```python
nltk.ne_chunk(postag)
```
