```python
words = ['eating','eat','eaten','writing','writes','programming','programs','finally','final','finallizing']
```

## PorterStemmer

```python
from nltk.stem  import PorterStemmer

stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"---->"+stemming.stem(word))
```

```python
# RegexpStemmer class --> regular expression stemmer algorithms.
# it basically taks a single regular expression and removes 
# any prefix or suffix that matches the expression
```

```python
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer('ing$|e$|s$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```
